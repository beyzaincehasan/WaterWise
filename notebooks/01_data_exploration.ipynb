```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)

print("WaterWise veri analizi ortamı hazır.")
```

```python
DATA_PATH = "../data/HSB_Living_Lab_Water_Consumption_Anonymized.csv"

df = pd.read_csv(
    DATA_PATH,
    sep=";",
    parse_dates=["timestamp"]
)

print("Veri seti yüklendi.")
print("Veri boyutu:", df.shape)
```

```python
df.head()
```

```python
print("Sütunlar:\n")

for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")
```

```python
df.info()
```

```python
print("Kayıt sayısı :", len(df))
print("Daire sayısı :", df["apartment"].nunique())
print("Sensör sayısı:", df["sensor_id"].nunique())

print("\nTarih aralığı:")
print("Başlangıç:", df["timestamp"].min())
print("Bitiş    :", df["timestamp"].max())
```

```python
missing = pd.DataFrame({
    "Eksik Sayısı": df.isnull().sum(),
    "Eksik Oranı (%)": (df.isnull().mean() * 100).round(2)
})

missing
```

```python
missing_apartment = df[df["apartment"].isna()]

print("Apartment bilgisi olmayan kayıt sayısı:",
      len(missing_apartment))

print("\nRoom type dağılımı:")
print(missing_apartment["room_type"].value_counts())

print("\nKullanım noktası dağılımı:")
print(missing_apartment["attached_to"].value_counts())
```

```python
missing_cluster = df[df["cluster_name"].isna()]

print("Cluster bilgisi olmayan kayıt sayısı:",
      len(missing_cluster))

print("\nApartment dağılımı:")
print(missing_cluster["apartment"].value_counts(dropna=False).head(20))

print("\nRoom type dağılımı:")
print(missing_cluster["room_type"].value_counts())
```

```python
missing_cluster = df[df["cluster_name"].isna()]

print("Cluster bilgisi olmayan kayıt sayısı:",
      len(missing_cluster))

print("\nApartment dağılımı:")
print(missing_cluster["apartment"].value_counts(dropna=False).head(30))

print("\nRoom type dağılımı:")
print(missing_cluster["room_type"].value_counts())

print("\nKullanım noktası dağılımı:")
print(missing_cluster["attached_to"].value_counts())
```

```python
df["value"].describe(
    percentiles=[0.25, 0.50, 0.75, 0.90, 0.95, 0.99]
)
```

```python
print("Minimum value :", df["value"].min())
print("Maximum value :", df["value"].max())

print("\n0 litre olan kayıt sayısı:",
      (df["value"] == 0).sum())

print("Negatif kayıt sayısı:",
      (df["value"] < 0).sum())
```

```python
total_liters = df["value"].sum()
total_m3 = total_liters / 1000

print(f"Toplam tüketim: {total_liters:,.2f} litre")
print(f"Toplam tüketim: {total_m3:,.2f} m³")
```

```python
fixture_stats = (
    df.groupby("attached_to")["value"]
      .agg(["count", "sum", "mean", "median", "max"])
      .sort_values("sum", ascending=False)
)

fixture_stats
```

```python
df["value_liters"] = df["value"] * 1000
```

```python
df[["value", "value_liters"]].head(10)
```

```python
total_liters = df["value_liters"].sum()

print(f"Toplam tüketim: {total_liters:,.2f} litre")
print(f"Toplam tüketim: {total_liters / 1000:,.2f} m³")
```

```python
print("Daire sayısı:", apartment_df["apartment"].nunique())
print()

print(
    apartment_df["apartment"]
    .value_counts()
    .sort_index()
)
```
